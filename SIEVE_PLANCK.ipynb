```python
import numpy as np

def generate_primes_skeleton(n):
    """
    Classical Sieve of Eratosthenes to isolate the prime skeleton.
    Establishes the discrete coordinate system for the AST framework.
    """
    sieve = [True] * (n + 1)
    for p in range(2, int(n**0.5) + 1):
        if sieve[p]:
            for i in range(p*p, n + 1, p):
                sieve[i] = False
    return [p for p in range(2, n + 1) if sieve[p]]

def compute_lefm_spectrum(primes, raw_phases, sigma):
    """
    Executes the complex-valued Laplace-Euler-Fourier-Mellin operator.
    Maps the prime coordinates to complex eigenvalues at a specified scale sigma.
    """
    r = np.log(primes)
    # Mellin scale invariance and Laplace decay envelope parameterized by sigma
    decay_envelope = np.exp(-0.1 * r) * (r ** (sigma - 1))

    # Synthesis of the Fourier phase modulation and complex representation
    eigenvalues = decay_envelope * (np.cos(raw_phases) + 1j * np.sin(raw_phases))
    return eigenvalues

def execute_unified_ast_engine(max_prime=200000, steps=10):
    """
    Unified verification execution kernel.
    Tracks the concurrent convergence of the fine-structure constant (alpha)
    and the emergence of the Planck action invariant (hbar) under Seed 123.
    """
    # Physical constants for baseline comparison
    alpha_physical = 1 / 137.035999

    print("=" * 90)
    print("         ARITHMETIC SPECTRAL THEORY (AST) UNIFIED UNIFICATION ENGINE")
    print("=" * 90)
    print(f"Isolating prime coordinates up to p = {max_prime}...")
    primes = generate_primes_skeleton(max_prime)
    total_primes = len(primes)
    print(f"Total prime coordinates anchored: {total_primes}")

    # Enforce the rigid deterministic seed configuration
    np.random.seed(123)
    raw_phases = np.random.uniform(0, 2 * np.pi, total_primes)
    print("Deterministic phase matrix initialized with Seed 123.\n")

    print(f"{'Horizon Limit':<14} | {'Count':<7} | {'Alpha_Arith':<12} | {'Alpha Delta':<12} | {'Planck Hbar':<12}")
    print("-" * 90)

    step_size = total_primes // steps
    d_sigma = 1e-5  # Infinitesimal probe for the Planck action derivative

    for i in range(1, steps + 1):
        current_limit = i * step_size
        if current_limit > total_primes:
            current_limit = total_primes

        # Segment the current horizon window
        sub_primes = primes[:current_limit]
        sub_phases = raw_phases[:current_limit]

        # 1. EVALUATE FINE-STRUCTURE COUPLING AT STATIONARITY (sigma = 0.5)
        spectrum_critical = compute_lefm_spectrum(sub_primes, sub_phases, sigma=0.5)

        integrated_phase = np.sum(np.angle(spectrum_critical))
        integrated_magnitude = np.sum(np.abs(spectrum_critical) ** 2)

        # Arithmetic fine-structure invariant
        alpha_arith = np.abs(integrated_phase / (integrated_magnitude * current_limit))
        alpha_delta = np.abs(alpha_arith - alpha_physical)

        # 2. EVALUATE PLANCK GRADIENT VIA SCALE PERTURBATION (0.5 +/- d_sigma)
        spectrum_up = compute_lefm_spectrum(sub_primes, sub_phases, sigma=0.5 + d_sigma)
        spectrum_down = compute_lefm_spectrum(sub_primes, sub_phases, sigma=0.5 - d_sigma)

        coherence_up = np.real(np.sum(spectrum_up))
        coherence_down = np.real(np.sum(spectrum_down))

        # Numerical calculation of the operational action stiffness (hbar)
        action_gradient = (coherence_up - coherence_down) / (2 * d_sigma)
        hbar_arith = np.abs(action_gradient / current_limit)

        max_p = sub_primes[-1]
        print(f"p <= {max_p:<9} | {current_limit:<7} | {alpha_arith:.8f}  | {alpha_delta:.8f}  | {hbar_arith:.8f}")

    print("=" * 90)
    print("Execution complete. All intermediate states fixed via deterministic kernel.")
    print("=" * 90)

if __name__ == "__main__":
    execute_unified_ast_engine(max_prime=200000, steps=10)
```
